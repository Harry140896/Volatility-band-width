```python
import json, base64
from pathlib import Path

nb = json.load(open("volatility_band_width.ipynb", encoding="utf-8"))
out = Path("figures"); out.mkdir(exist_ok=True)
k = 0
for cell in nb["cells"]:
    for o in cell.get("outputs", []):
        png = o.get("data", {}).get("image/png")
        if png:
            k += 1
            (out / f"figure_{k:02d}.png").write_bytes(base64.b64decode(png))
print(k, "figures saved")
```
